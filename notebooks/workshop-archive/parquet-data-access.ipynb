## Authenticate with HydroShare

````python
from utils import S3hsclient
````

````python
hs = S3hsclient.S3HydroShare()
````

We want to access the STREAM data stored in Parquet format on HydroShare, available at https://hydroshare.org/resource/4553ef8a14884e4199c7018975b8faaa/. To do this, we'll use the HydroShare Python library.

Parquet is a columnar, compressed file format: it stores data by column rather than by row and lets us read only the specific columns and row groups we need. This makes it much more efficient than row-based formats like CSV, especially for a large dataset like STREAM, where we often only need a subset of variables or sites.

````python
resource_id = '4553ef8a14884e4199c7018975b8faaa'
resource = hs.resource(resource_id)
````

This resource contains a number of files that can be loaded for analysis. To see which files are available we'll use the `s3_ls` command.

````python
resource.s3_ls()
````

You may notice there are different types of content items in this resource: some end with `.parquet` (a single file), while others have no extension at all. Those without an extension are actually directories — deeper structures containing one or more Parquet files, not a single file. Let's check the structure of both kinds, starting with a `.parquet` item:

````python
resource.s3_ls('watersheds.parquet')
````

The output above shows that `watersheds.parquet` returned only itself — this confirms it's a single file, not a directory: there's nothing "inside" it to list.

Now let's try the other kind: `streamflow`. The following command lists its immediate subfolders:

````python
resource.s3_ls('streamflow/')[:15]
````

The command above is useful, but it only lists one level at a time — to see everything nested inside `streamflow`, we'd need to call `s3_ls()` repeatedly on each subfolder it returns. A simpler way is to use `find()`, which recursively lists every file underneath in a single call. This is the key command, since it reveals the actual naming pattern used to organize the data:

````python
fs = hs.get_s3_filesystem()
fs.find('tonycastronova/4553ef8a14884e4199c7018975b8faaa/data/contents/streamflow')
````

From the output above, you can see that `STREAM_ID=STREAM-gauge-1014` is split across two Parquet files. This directory naming pattern (`key=value`) is called **Hive partitioning**: `STREAM_ID` is the partition key, and each gauge gets its own subfolder named after its value. This structure is what lets tools like `pandas` and `dask` filter directly on `STREAM_ID` without having to scan every file in the dataset.

## Accessing Data using Scientific Python Libraries

Load a subset of STREAM dataset for a gage of interest and inspect the data. This can be done using several common scientific Python libraries.

<a id="pandas-dataframe"></a>
### Pandas DataFrame
*An example of accessing streamflow data.*

````python
import pandas
````

Let's load the metadata table and use it to choose a gage of interest.

````python
metadata_df = pandas.read_parquet(
    'tonycastronova/4553ef8a14884e4199c7018975b8faaa/data/contents/metadata.parquet',
    filesystem=hs.get_s3_filesystem()
)

metadata_df
````

The map below plots every gage in `metadata_df`. Click a point to see that gage's full metadata in a popup — useful for browsing the catalog and picking a gage to explore in the rest of this notebook.

````python
from utils import gage_metadata_map
gage_metadata_map.plot_gage_metadata_map(metadata_df)
````

Based on the map above (or by filtering `metadata_df` directly), set `gage_id` to the STREAM gage you'd like to explore for the rest of this notebook:

````python
gage_id = 'STREAM-gauge-1022'
````

The following example loads the entire record of streamflow data. 

````python
%%time

streamflow_path = f'tonycastronova/4553ef8a14884e4199c7018975b8faaa/data/contents/streamflow'

df = pandas.read_parquet(
    f'{streamflow_path}/STREAM_ID={gage_id}',
    filesystem=hs.get_s3_filesystem()
)

df
````

Show a summary of record count, date span in both days and years, min/max/mean/std of Q_m3s, and a quick count of how many records fall under each flag (A, etc.).

````python
flags = sorted(df["Flag_Q_m3s"].unique())

summary = pandas.Series({
    "records": len(df),
    "start_date": df["DateTime"].min(),
    "end_date": df["DateTime"].max(),
    "period_days": (df["DateTime"].max() - df["DateTime"].min()).days,
    "period_years": round((df["DateTime"].max() - df["DateTime"].min()).days / 365.25, 1),
    "n_unique_flags": len(flags),
    "flags": ", ".join(flags),
    "Q_min_m3s": df["Q_m3s"].min(),
    "Q_max_m3s": df["Q_m3s"].max(),
    "Q_mean_m3s": round(df["Q_m3s"].mean(), 3),
    "Q_std_m3s": round(df["Q_m3s"].std(), 3),
}, name="value").to_frame()

summary
````

Plot the streamflow for one of the STREAM sites.

````python
import matplotlib.pyplot as plt

plot_df = df.set_index('DateTime')

fig, ax = plt.subplots(figsize=(12, 4))
plot_df['Q_m3s'].plot(ax=ax, color='steelblue', linewidth=0.8)

ax.set_title(f'Streamflow for {gage_id}')
ax.set_xlabel('Date')
ax.set_ylabel('Streamflow (m³/s)')
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
````

Up to this point, we haven't downloaded any data to disk. Everything we've worked with (metadata_df, df) has stayed in memory. This is one of the key advantages of Parquet, and cloud-native data formats in general: you can read exactly the columns and rows you need directly from cloud storage into memory, without downloading and storing the entire dataset locally first.

If you'd like a local copy of the data you just plotted for gage_id, you can save it to disk with the following code:

```python
output_file = f'{gage_id}_streamflow.csv'
df.to_csv(output_file, index=False)
print(f'Data saved to {output_file}')
```

### Dask DataFrame
*An example of accessing multiple water quality parameters.*

The advantage of using a Dask DataFrame instead of a Pandas DataFrame is that it provides lazy loading and delayed computations. This is especially beneficial when analyzing large amounts of data.

````python
import dask.dataframe as dd
````

Just like with `streamflow`, let's first explore the structure of the `waterquality_timeseries` folder to confirm it follows the same Hive-partitioned layout before reading from it:

````python
fs = hs.get_s3_filesystem()
wq_path = f'tonycastronova/4553ef8a14884e4199c7018975b8faaa/data/contents/waterquality_timeseries'
fs.find(wq_path)[:15]
````

Now that we've confirmed `waterquality_timeseries` follows the same `STREAM_ID=...` partitioning as `streamflow`, let's read this gage's water quality data using Dask:

````python
# create a dataset pointing to the files
ds = dd.read_parquet(
    f'{wq_path}/STREAM_ID={gage_id}',
    filesystem=hs.get_s3_filesystem()
)

ds
````

Notice how fast this query was. That's because no data has been downloaded yet. The `ds` dataset is just a lazy plan, and Dask only pulls data from HydroShare once we call `.compute()`.

Unlike streamflow, which has a single variable (`Q_m3s`), water quality data includes many parameters — temperature, nitrate, dissolved oxygen, pH, turbidity, chlorophyll, and more.

Most gauges only report a handful of these (see the `WQ_parameters` column in `metadata_df`), so selecting just the columns you need matters even more here, since it avoids pulling in columns that may be entirely empty for this gauge. Below, we'll look up which parameters are actually available for this gage and select two of them before computing.

In this basic example we're already scoped to a single gage's partition, so this only demonstrates column selection. Dask's real efficiency shows up when querying across many gages at once. By pointing at the top-level `waterquality_timeseries` directory and filtering on `STREAM_ID`, Dask can use Hive partitioning to skip entire gage folders without reading any of their files.

Let's look up which water quality parameters are actually available for this gage — not every gage measures the same variables:

````python
wq_params_str = metadata_df.loc[metadata_df.STREAM_ID == gage_id, 'WQ_parameters'].values[0]
wq_params = [p.strip() for p in wq_params_str.split(',')]

from IPython.display import HTML

badges = "".join(
    f"<span style='display:inline-block; margin:3px; padding:4px 10px; "
    f"background:#e8f0fe; color:#1a56db; border-radius:12px; font-size:13px;'>{p}</span>"
    for p in wq_params
)
HTML(f"<b>Water quality parameters available for {gage_id}:</b><br>{badges}")
````

Now choose two parameters to compare — feel free to swap these for any two shown above:

````python
param_x = 'WTemp_C'  # water temperature in degrees Celsius
param_y = 'DO_mgL'   # dissolved oxygen in mg/L
````

Water quality parameters can have independent gaps in time, so before plotting we need to keep only the records where both chosen parameters actually have valid values at the same timestamp:

````python
wq_df = ds[['DateTime', param_x, param_y]].compute()

overlap_df = wq_df.dropna(subset=[param_x, param_y])

print(f'{len(overlap_df)} overlapping records out of {len(wq_df)} total')
````

With the overlapping subset in hand, we can now create a scatter plot comparing the two parameters:

````python
# plot the data
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 6))
ax.scatter(overlap_df[param_x], overlap_df[param_y], alpha=0.5, s=15, color='steelblue', edgecolor='none')

ax.set_title(f'{param_y} vs {param_x} for {gage_id}')
ax.set_xlabel(param_x)
ax.set_ylabel(param_y)
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
````
